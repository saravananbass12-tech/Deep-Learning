```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Input
```

```python
data=pd.read_csv("ann_mood_thinking_dataset_27000.csv")
data
```

```python
data.columns = data.columns.str.strip()
```

```python
x= data.drop(columns=[
    'record_id',
    'mood',
    'thinking_pattern'
])

x= pd.get_dummies(x, dtype=int)
```

```python
mood_labels = {
    'Happy': 0,
    'Sad': 1,
    'Angry': 2,
    'Stressed': 3
}

y_mood = data['mood'].map(mood_labels).values
```

```python
thinking_labels = {
    'Logical': 0,
    'Emotional': 1,
    'Creative': 2,
    'Analytical': 3
}

y_thinking = data['thinking_pattern'].map(thinking_labels).values
```

```python
scaler = StandardScaler()
x_scaled = scaler.fit_transform(x)
```

```python
model=Sequential()
model.add(Input(shape=(x_scaled.shape[1],)))
model.add(Dense(16,activation='relu'))
model.add(Dense(8,activation='relu'))
model.add(Dense(4,activation='softmax'))
```

```python
model.compile(
    loss='sparse_categorical_crossentropy',
    optimizer='adam',
    metrics=['accuracy']
)
```

```python
model2 = Sequential()
model2.add(Input(shape=(x_scaled.shape[1],)))
model2.add(Dense(16, activation='relu'))
model2.add(Dense(8, activation='relu'))
model2.add(Dense(4, activation='softmax'))
```

```python
model2.compile(
    loss='sparse_categorical_crossentropy',
    optimizer='adam',
    metrics=['accuracy']
)
```

```python
model.fit(x, y_mood, epochs=50)
```

```python
model2.fit(x, y_thinking, epochs=50)
```

```python
age = int(input("Enter Age: "))
sleep = float(input("Enter Sleep Hours: "))
stress = int(input("Enter Stress Level (1-10): "))
energy = int(input("Enter Energy Level (1-10): "))
social = int(input("Enter Social Interaction (1-10): "))
focus = int(input("Enter Focus Level (1-10): "))
creativity = int(input("Enter Creativity Score: "))
logic = int(input("Enter Logic Score: "))
sensitivity = int(input("Enter Emotional Sensitivity: "))
decision = input("Enter Decision Speed: ")
problem = input("Enter Problem Solving Style: ")
activity = input("Enter Recent Activity: ")

```

```python
test = pd.DataFrame([{
    'age': age,
    'sleep_hours': sleep,
    'stress_level_1_to_10': stress,
    'energy_level_1_to_10': energy,
    'social_interaction_1_to_10': social,
    'focus_level_1_to_10': focus,
    'creativity_score_1_to_10': creativity,
    'logic_score_1_to_10': logic,
    'emotional_sensitivity_1_to_10': sensitivity,
    'decision_speed': decision,
    'problem_solving_style': problem,
    'recent_activity': activity
}])

test = pd.get_dummies(test, dtype=int)

test = test.reindex(columns=pd.get_dummies(
    data.drop(columns=['record_id', 'mood', 'thinking_pattern']),
    dtype=int
).columns, fill_value=0)

test = scaler.transform(test)
```

```python
pred = model.predict(test, verbose=0)
mood_names = ['Happy', 'Sad', 'Angry', 'Stressed']
print("Predicted Mood:", mood_names[np.argmax(pred)])

pred2 = model2.predict(test, verbose=0)
thinking_names = ['Logical', 'Emotional', 'Creative', 'Analytical']
print("Thinking Style:", thinking_names[np.argmax(pred2)])
```
